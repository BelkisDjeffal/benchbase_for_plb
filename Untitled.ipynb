```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

# Load the CSV file
df = pd.read_csv("summary_results_new.csv")

# Convert Latency values from microseconds to milliseconds
df["Average Latency"] /= 1000
df["95th Percentile Latency"] /= 1000
df["Maximum Latency"] /= 1000
df["Median Latency"] /= 1000
df["Minimum Latency"] /= 1000
df["25th Percentile Latency"] /= 1000
df["90th Percentile Latency"] /= 1000
df["99th Percentile Latency"] /= 1000
df["75th Percentile Latency"] /= 1000

# Extract Premium and Freemium Metrics
df["Premium:Freemium"] = df["Terminal Classes Weights"].astype(str)

# Define function for grouped bar chart
def plot_grouped_bar_chart(df, metric, title, ylabel, output_name):
    fig, ax = plt.subplots(figsize=(10, 6))
    width = 0.35
    x = np.arange(len(df))

    ax.bar(x - width/2, df[f"premium_{metric}"], width, label="Premium")
    ax.bar(x + width/2, df[f"freemium_{metric}"], width, label="Freemium")

    ax.set_xlabel("Clients (Total, Premium:Freemium Split)")
    ax.set_ylabel(ylabel)
    ax.set_title(title)
    ax.set_xticks(x)
    ax.set_xticklabels(df["Premium:Freemium"], rotation=45, ha="right")
    ax.legend()

    # Save & Show Plot
    plt.grid(axis="y")
    plt.tight_layout()
    plt.savefig(f"{output_name}.png", dpi=300)
    plt.savefig(f"{output_name}.pdf", format="pdf")
    plt.show()

# Plot Average Latency
plot_grouped_bar_chart(
    df,
    metric="Average Latency",
    title="Average Latency vs. Premium:Freemium Breakdown",
    ylabel="Average Latency (milliseconds)",
    output_name="average_latency"
)

# Plot 95th Percentile Latency
plot_grouped_bar_chart(
    df,
    metric="95th Percentile Latency",
    title="95th Percentile Latency vs. Premium:Freemium Breakdown",
    ylabel="95th Percentile Latency (milliseconds)",
    output_name="latency_95"
)

# Plot Throughput
plot_grouped_bar_chart(
    df,
    metric="Throughput",
    title="Throughput vs. Premium:Freemium Breakdown",
    ylabel="Throughput (requests/second)",
    output_name="throughput"
)

# Plot Goodput
plot_grouped_bar_chart(
    df,
    metric="Goodput",
    title="Goodput vs. Premium:Freemium Breakdown",
    ylabel="Goodput (requests/second)",
    output_name="goodput"
)

# Plot Attempted Throughput
plot_grouped_bar_chart(
    df,
    metric="AttemptedThroughput",
    title="Attempted Throughput vs. Premium:Freemium Breakdown",
    ylabel="Attempted Throughput (requests/second)",
    output_name="attempted_throughput"
)
```
