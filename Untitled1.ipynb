```python

# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load CSV Data
csv_file = 'summary_results_new.csv' 
df = pd.read_csv(csv_file)

# Create a column for terminal configuration (e.g., '15:35')
df['Terminal Config'] = df['Terminal Classes Weights'].astype(str)

# Scatter Plot for Throughput
plt.figure(figsize=(12, 6))
sns.scatterplot(x='Terminal Config', y='premium_Throughput', data=df, label='Premium', color='blue', s=100)
sns.scatterplot(x='Terminal Config', y='freemium_Throughput', data=df, label='Freemium', color='orange', s=100)
plt.xlabel('Terminals Configuration', fontsize=12)
plt.ylabel('Throughput (req/s)', fontsize=12)
plt.title('Scatter Plot of Throughput per Terminal Configuration', fontsize=14)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Scatter Plot for Goodput
plt.figure(figsize=(12, 6))
sns.scatterplot(x='Terminal Config', y='premium_Goodput', data=df, label='Premium', color='green', s=100)
sns.scatterplot(x='Terminal Config', y='freemium_Goodput', data=df, label='Freemium', color='red', s=100)
plt.xlabel('Terminal Configuration', fontsize=12)
plt.ylabel('Goodput (req/s)', fontsize=12)
plt.title('Scatter Plot of Goodput per Terminal Configuration', fontsize=14)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Scatter Plot for Average Latency
plt.figure(figsize=(12, 6))
sns.scatterplot(x='Terminal Config', y='Average Latency', data=df, label='Average Latency', color='purple', s=100)
plt.xlabel('Terminal Configuration', fontsize=12)
plt.ylabel('Average Latency (microseconds)', fontsize=12)
plt.title('Scatter Plot of Average Latency per Terminal Configuration', fontsize=14)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
